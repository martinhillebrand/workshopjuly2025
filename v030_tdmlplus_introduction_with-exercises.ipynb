# Getting sample data

````python
from sklearn.datasets import make_classification
import pandas as pd
import numpy as np

X, y = make_classification(
    n_samples=1000,
    n_features=6,
    n_informative=4,
    n_redundant=2,
    n_clusters_per_class=1,
    class_sep=1.0,
    random_state=42
)

df = pd.DataFrame(X, columns=[f"feature_{i}" for i in range(X.shape[1])])
df["target"] = y

df = df.reset_index().rename(columns = {"index":"row_id"})

tdml.copy_to_sql(df, "toydata",primary_index="row_id", if_exists="replace")
tdml.copy_to_sql(df.rename(columns= {"feature_0":"feature_20"}), "toydata2",primary_index="row_id", if_exists="replace")
# sample series
s = pd.Series(['a', 'b', 'c', 'a', 'd'])

# mapping dictionary
mapping = {'a': 1, 'b': 2, 'c': 3}

# apply map
mapped = s.map(mapping)

#print(mapped)
dfx = pd.DataFrame({"mycol":s}).reset_index().rename(columns={"index":"row_id"})
tdml.copy_to_sql(dfx,"maptest", primary_index="row_id", if_exists="replace")
````

````python
DF = tdml.DataFrame("toydata")
DF2 = tdml.DataFrame("toydata2")
DF_mt = tdml.DataFrame("maptest")
````

````python
DF_new_cols = DF3.assign(
    #trycast
    new_feat_trycast = DF3.newfeat.___(tdml.DECIMAL(9,6)),  # Replace with the method to attempt casting a column to a new type.
    
    #hashbin
    f0hash = DF3.feature_0.___(3),  # Replace with the method to hash a column.
    f0hash2 = DF3.feature_0.___(3),  # Replace with the method to hash a column.
    f0hash_salt = DF3.feature_0.___(3, "salt"),  # Replace with the method to hash a column with a salt.
    f3hash = DF3.feature_0.___(100),  # Replace with the method to hash a column.
    
    # power transform
    feature_0_pt = DF3.feature_0.___(),  # Replace with the method to apply a power transformation.
    feature_0_pt_bx = DF3.feature_0.___('box-cox'),  # Replace with the method to apply a power transformation with a specific method.

)
DF_new_cols
````
